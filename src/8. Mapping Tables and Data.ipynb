**4a. Gold Table - Subject Summary**

```python
%sql
select SUBJID, TOTAL_AES, SERIOUS_AES, FATAL_AES, TOTAL_AES from cluepoints_db_demo.gold.subject_summary limit 10
```

**4b. Gold Table - Country Summary**

```python
%sql
SELECT 
    COUNTRY,
    COUNT(DISTINCT SUBJID) as TOTAL_SUBJECTS,
    SUM(TOTAL_AES) as TOTAL_AES,
    SUM(SERIOUS_AES) as SERIOUS_AES,
    SUM(FATAL_AES) as FATAL_AES,
    ROUND(SUM(TOTAL_AES) / COUNT(DISTINCT SUBJID), 2) as AVG_AES_PER_SUBJECT
FROM cluepoints_db_demo.gold.subject_summary
GROUP BY COUNTRY
ORDER BY TOTAL_SUBJECTS DESC;
```

**4c. Gold Table - Site and Country Summary**

```python
%sql
SELECT 
    SITEID,
    COUNTRY,
    COUNT(DISTINCT SUBJID) as TOTAL_SUBJECTS,
    SUM(SERIOUS_AES) as SERIOUS_AES,
    ROUND(SUM(SERIOUS_AES) * 100.0 / COUNT(DISTINCT SUBJID), 1) as SERIOUS_AE_RATE_PCT
FROM cluepoints_db_demo.gold.subject_summary
GROUP BY SITEID, COUNTRY
ORDER BY SERIOUS_AE_RATE_PCT DESC;
```

```python
%sql
SELECT 
    ARM,
    COUNT(DISTINCT SUBJID) as TOTAL_SUBJECTS,
    SUM(SERIOUS_AES) as SERIOUS_AES,
    ROUND(SUM(SERIOUS_AES) * 100.0 / COUNT(DISTINCT SUBJID), 1) as SERIOUS_AE_RATE_PCT
FROM cluepoints_db_demo.gold.subject_summary
GROUP BY ARM
ORDER BY SERIOUS_AE_RATE_PCT DESC;
```
